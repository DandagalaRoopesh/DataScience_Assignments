```python
#Hospital billing system
room_charge_per_day = 2000
doctor_fee = 1500
lab_test = 300

days_stayed = 4
lab_tests = 3
medicine_charges = 2400

room_charges = days_stayed * room_charge_per_day
lab_charges = lab_tests * lab_test

total_charges = room_charges + lab_charges + doctor_fee + medicine_charges

if total_charges > 10000:
    discount = total_charges * 0.10
else:
    discount = 0

final_bill = total_charges - discount

print("Room Charges:", room_charges)
print("Doctor Fee:", doctor_fee)
print("Lab Charges:", lab_charges)
print("Medicine Charges:", medicine_charges)
print("Total Before Discount:", total_charges)
print("Discount Applied:", discount)
print("Final Bill:", final_bill)
```
