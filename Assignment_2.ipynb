```python
#1.	Create a list of 5 integers. Perform and print the result of the following operations: 
# append, extend, insert, remove, pop, clear, index, count, sort, and reverse
```

```python
List = [1,2,3,4,5] # created a list of 5 integers
print(List)
```

```python
# append : Append is used to add the elements into a list , when append is used the new element is added to the end of the list.

List.append(6)
print(List) 
```

```python
# extend : The list.extend() method is used to append all elements of an iterable (like another list) to the end of the current list. 
#It modifies the original list in place and does not return a new list.

List_a = [1,2,3]
List_b = [4,5,6]

List_a.extend(List_b)
print(List_a)
```

```python
# insert : the List.insert() methos id used to add new elements at any position, 
# need to provide the index value where the new element need to be inserted

List.insert(3,7)
print(List)
```

```python
# remove : it is used to remove a specific value from a list 
# he remove() method deletes only the first occurrence of the value you specify.
List = [26,36,3,56,76]
List.remove(3)
print(List)
```

```python
# pop : The pop() method removes the last item in a list, but it lets you work 
# with that item after removing it.
List= [15,20,50,36,70,50,100]
List.pop()
print(List)
```

```python
# clear : The clear() method removes all the elements from a list.

List = [1,2,3,4,5,6]
print(List)

List.clear()
print(List)
```

```python
# index : To find the position of an element in a Python list, the most common approach is 
#to use the built-in list.index() method, 
#which returns the index of the first occurrence of the specified value.

List = [10,20,25,35,40,60,56,78,96,45,50]
print(List)
print(List.index(45))
```

```python
# count : The count() method in Python is used to count the number of occurrences of a specified value in a list or string.

List = [10,20,25,35,40,25,60,56,78,96,45,25,50]
print(List)
count_list = List.count(25)
print(count_list)
```

```python
# sort : The sort() method in Python sorts the elements of a list in ascending order by default. 
#It can also sort values in descending order or based on a custom function.
List = [10,20,25,35,40,25,60,56,78,96,45,25,50]
print(List)

List.sort()
print(List)
```

```python
# reverse :  .reverse() method on the list to reverse it in place. Example: list_name.reverse(). This modifies the original list.

List = [10,20,25,35,40,25,60,56,78,96,45,25,50]
print(List)

List.reverse()
print(List)

# by using sorting
List.sort()
print(List)

List.sort(reverse = True)
print(List)
```

```python
"""
6.	Given a 'marks' variable, print the grade:
- 90 and above: 'A'
- 75–89: 'B'
- 50–74: 'C'
- Below 50: 'Fail'

"""

Marks = int(input("Enter Marks: "))

if Marks >= 90:
    print("Grade : A")
elif Marks >= 75 and Marks <= 89:
    print("Grade : B")
elif Marks >= 50 and Marks <= 74:
    print("Grade : C")
else:
    print("Grade : Fail")
    
```

```python
"""
11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows

"""
import pandas as pd
import numpy as np

# creating dataframe manually

data = {
    'Name': ['Ravi', 'Anjali', 'Kiran', 'Sneha', 'Amit', 
             'Ravi', 'Pooja', 'Vikram', 'Neha', 'Kiran'],
    'Age': [25, 30, 28, 22, 35, 25, 27, 31, 29, 28],
    'City': ['Hyderabad', 'Mumbai', 'Delhi', 'Chennai', 'Pune',
             'Hyderabad', 'Bangalore', 'Delhi', 'Mumbai', 'Delhi'],
    'Salary': [50000, 60000, np.nan, 45000, 70000,
               50000, 52000, np.nan, 61000, np.nan]
}

df = pd.DataFrame(data)

print("Original DataFrame")
print(df)
```

```python
# info()
print("\nInfo:")
print(df.info())
```

```python
# describe()
print("\ndescribe")
print(df.describe())
```

```python
#select name and city columns
print("\nselecting name and city")
print(df[['Name','City']])
```

```python
# Drop 'City' column

df = df.drop('City', axis = 1)
print("\nAfter dropping the city column")
print(df)
```

```python

#  Fill any missing values in 'Salary' column with the mean

mean_salary = df['Salary'].mean()
df['Salary'] = df['Salary'].fillna(mean_salary)

print(df)
```

```python
# Remove any duplicate rows

df = df.drop_duplicates()
print(df)
```
