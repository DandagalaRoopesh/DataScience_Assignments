**2.Split the dataset into training and testing sets (e.g., 80% training, 20% testing).**

```python
# =============================================================================
# DATASET SPLITTING FOR MULTIPLE LINEAR REGRESSION
# =============================================================================

# Multiple Linear Regression requires two types of variables:
#
# 1. Independent Variables (Predictors)
#    These variables are used to predict the target variable.
#
# Examples:
# Age_08_04, KM, HP, Weight, Fuel_Type_Diesel, etc.
#
# 2. Dependent Variable (Target Variable)
#    This is the variable we want to predict.
#
# Example:
# Price
#
# Therefore, we first separate the target variable
# from the predictor variables.

# Independent Variables (Features)
X = df.drop('Price', axis=1)

# Dependent Variable (Target)
y = df['Price']
```

```python
# =============================================================================
# IMPORT TRAIN TEST SPLIT FUNCTION
# =============================================================================

# train_test_split is imported from sklearn.model_selection.
#
# It is used to divide the dataset into:
#
# 1. Training Data
# 2. Testing Data
#
# Training Data:
# Used to train the regression model.
#
# Testing Data:
# Used to evaluate model performance on unseen data.

from sklearn.model_selection import train_test_split
```

```python
# =============================================================================
# SPLITTING THE DATASET
# =============================================================================

# Split the dataset into training and testing datasets.
#
# test_size = 0.20
# Means:
# 20% of data will be used for testing.
# 80% of data will be used for training.
#
# random_state = 42
# Ensures reproducibility.
# Every time the code runs,
# the same train-test split will be generated.

X_train, X_test, y_train, y_test = train_test_split(
    
    X,
    y,
    
    test_size=0.20,
    
    random_state=42
    
)
```

```python
# =============================================================================
# VERIFYING TRAINING AND TESTING DATASET DIMENSIONS
# =============================================================================

# Display dimensions of training and testing datasets.

print("Training Feature Shape :", X_train.shape)

print("Testing Feature Shape  :", X_test.shape)

print("Training Target Shape  :", y_train.shape)

print("Testing Target Shape   :", y_test.shape)
```

```python
# =============================================================================
# DISPLAY TRAINING AND TESTING RECORD COUNTS
# =============================================================================

print("Total Records     :", len(df))

print("Training Records  :", len(X_train))

print("Testing Records   :", len(X_test))
```

```python
# =============================================================================
# VISUALIZING TRAINING AND TESTING SPLIT
# =============================================================================

split_data = {
    'Training Data': len(X_train),
    'Testing Data': len(X_test)
}

plt.figure(figsize=(6,5))

plt.pie(
    split_data.values(),
    labels=split_data.keys(),
    autopct='%1.1f%%'
)

plt.title("Training vs Testing Dataset Split")

plt.show()
```

Conclusion:

The dataset was successfully divided into
training and testing datasets using the
train_test_split() function from Scikit-Learn.

The predictor variables (X) and target variable (y)
were separated prior to splitting.

An 80:20 split ratio was used, where:

• 80% of the data was allocated for training.
• 20% of the data was allocated for testing.

The training dataset will be used to train the
Multiple Linear Regression model, while the testing
dataset will be used to evaluate model performance
on unseen data.

This step ensures an unbiased assessment of the
model's predictive capability and helps prevent
overfitting.

```python
# =============================================================================
# MODEL 1 : FULL MULTIPLE LINEAR REGRESSION MODEL
# =============================================================================

# Add constant term.
#
# In regression equations:
#
# Y = β0 + β1X1 + β2X2 + ...
#
# β0 represents the intercept.
#
# statsmodels does not add intercept automatically,
# therefore we add it manually.

X_train_const = sm.add_constant(X_train)

# Build the regression model.

model_1 = sm.OLS(
    y_train,
    X_train_const
).fit()

# Display complete statistical summary.

print(model_1.summary())
```

**4.Evaluate the performance of the model using appropriate evaluation metrics on the testing dataset**

```python
# =============================================================================
# IMPORTING LIBRARIES FOR MODEL EVALUATION
# =============================================================================

# These metrics are used to evaluate the performance
# of the Multiple Linear Regression model on unseen data.

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
```

```python
# =============================================================================
# GENERATING PREDICTIONS ON TESTING DATA
# =============================================================================

# The model has already learned patterns from the
# training dataset.

# Now we use the trained model to predict vehicle prices
# for the testing dataset.

# Add intercept term to testing dataset because
# statsmodels requires the same structure used during training.

X_test_const = sm.add_constant(X_test)

# Generate predicted values.

y_pred = model_1.predict(X_test_const)

# Display first few predictions.

print(y_pred.head())
```

```python
# =============================================================================
# MEAN ABSOLUTE ERROR (MAE)
# =============================================================================

# MAE calculates the average absolute difference
# between actual and predicted values.

mae = mean_absolute_error(
    y_test,
    y_pred
)

print("Mean Absolute Error (MAE):", mae)
```

```python
# =============================================================================
# MEAN SQUARED ERROR (MSE)
# =============================================================================

# MSE calculates the average squared difference
# between actual and predicted values.

mse = mean_squared_error(
    y_test,
    y_pred
)

print("Mean Squared Error (MSE):", mse)
```

```python
# =============================================================================
# ROOT MEAN SQUARED ERROR (RMSE)
# =============================================================================

# RMSE converts MSE back into the original unit
# of the target variable.

rmse = np.sqrt(mse)

print("Root Mean Squared Error (RMSE):", rmse)
```

```python
# =============================================================================
# R-SQUARED SCORE
# =============================================================================

# R-Squared measures how much variation
# in Price is explained by the model.

r2 = r2_score(
    y_test,
    y_pred
)

print("R-Squared Score:", r2)
```

```python
# =============================================================================
# MODEL PERFORMANCE SUMMARY
# =============================================================================

evaluation_results = pd.DataFrame({

    "Metric": [
        "MAE",
        "MSE",
        "RMSE",
        "R-Squared"
    ],

    "Value": [
        mae,
        mse,
        rmse,
        r2
    ]

})

print(evaluation_results)
```

**5.Apply Lasso and Ridge methods on the model.**

```python
# =============================================================================
# IMPORTING LIBRARIES FOR RIDGE AND LASSO REGRESSION
# =============================================================================

# Ridge Regression and Lasso Regression are regularization
# techniques used to improve model performance and reduce
# overfitting.

from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso

# Metrics used for evaluating model performance.

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
```

```python
# =============================================================================
# FEATURE SCALING
# =============================================================================

# Ridge and Lasso are sensitive to variable scales.
#
# Example:
#
# KM      -> 50000
# Doors   -> 5
#
# Without scaling, KM dominates the model.

from sklearn.preprocessing import StandardScaler

# Create scaler object.

scaler = StandardScaler()

# Fit scaler on training data and transform.

X_train_scaled = scaler.fit_transform(X_train)

# Apply same transformation to testing data.

X_test_scaled = scaler.transform(X_test)
```

```python
# =============================================================================
# RIDGE REGRESSION MODEL
# =============================================================================

# Create Ridge Regression model.
#
# alpha = Regularization Strength
#
# Higher alpha:
# More penalty
#
# Lower alpha:
# Less penalty

ridge_model = Ridge(alpha=1.0)

# Train model using training dataset.

ridge_model.fit(
    X_train_scaled,
    y_train
)

# Generate predictions on testing dataset.

ridge_predictions = ridge_model.predict(
    X_test_scaled
)
```

```python
# =============================================================================
# RIDGE MODEL EVALUATION
# =============================================================================

ridge_mae = mean_absolute_error(
    y_test,
    ridge_predictions
)

ridge_mse = mean_squared_error(
    y_test,
    ridge_predictions
)

ridge_rmse = np.sqrt(ridge_mse)

ridge_r2 = r2_score(
    y_test,
    ridge_predictions
)

print("RIDGE REGRESSION RESULTS")
print("------------------------")
print("MAE :", ridge_mae)
print("MSE :", ridge_mse)
print("RMSE:", ridge_rmse)
print("R²  :", ridge_r2)
```

```python
# =============================================================================
# RIDGE COEFFICIENT ANALYSIS
# =============================================================================

ridge_coefficients = pd.DataFrame({

    "Feature": X_train.columns,
    "Coefficient": ridge_model.coef_

})

print(ridge_coefficients)
```

```python
# =============================================================================
# LASSO REGRESSION MODEL
# =============================================================================

# Create Lasso Regression model.

lasso_model = Lasso(
    alpha=1.0,
    max_iter=10000
)

# Train model.

lasso_model.fit(
    X_train_scaled,
    y_train
)

# Generate predictions.

lasso_predictions = lasso_model.predict(
    X_test_scaled
)
```

```python
# =============================================================================
# LASSO MODEL EVALUATION
# =============================================================================

lasso_mae = mean_absolute_error(
    y_test,
    lasso_predictions
)

lasso_mse = mean_squared_error(
    y_test,
    lasso_predictions
)

lasso_rmse = np.sqrt(
    lasso_mse
)

lasso_r2 = r2_score(
    y_test,
    lasso_predictions
)

print("LASSO REGRESSION RESULTS")
print("------------------------")
print("MAE :", lasso_mae)
print("MSE :", lasso_mse)
print("RMSE:", lasso_rmse)
print("R²  :", lasso_r2)
```

```python
# =============================================================================
# LASSO COEFFICIENT ANALYSIS
# =============================================================================

lasso_coefficients = pd.DataFrame({

    "Feature": X_train.columns,
    "Coefficient": lasso_model.coef_

})

print(lasso_coefficients)
```

```python
# =============================================================================
# MODEL COMPARISON
# =============================================================================

comparison = pd.DataFrame({

    "Model": [

        "Multiple Linear Regression",
        "Ridge Regression",
        "Lasso Regression"

    ],

    "R² Score": [

        r2,
        ridge_r2,
        lasso_r2

    ],

    "RMSE": [

        rmse,
        ridge_rmse,
        lasso_rmse

    ]

})

print(comparison)
```
